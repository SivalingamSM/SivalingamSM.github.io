## 7. What is a DataFrame?

A **DataFrame** is a two-dimensional labelled data structure with rows and columns.

It is similar to a spreadsheet, an Excel table, or a database table.

Different columns may contain different data types.

A common way to create a DataFrame is from a dictionary.

### Question 8

Create the following DataFrame:

| Name | Age | Marks |
|---|---:|---:|
| A | 20 | 75 |
| B | 21 | 85 |
| C | 19 | 90 |


````python
data = {
    'Name': ['A', 'B', 'C','D','A', 'B', 'C','D'],
    'Age': [20, 21, 19, 22, 20, 21, 19, 22],
    'Marks': [75, 85, 90, 40, 75, 85, 90, 40]
}
df = pd.DataFrame(data)

print(df)
````

````python
df['Age']
````

````python
df[df['Age']>20]
````

````python
df.loc[(df['Age']>20) & (df['Marks']>80), "Name"]
````

````python
df["Name"] == "B"
````

````python
df.iloc[df["Name"] == "B"]
````

````python
df.iloc[(df["Name"] == "B").tolist()]
````

````python
df.loc[(df["Name"] == "B") | (df["Name"] == "C"), "Marks"]
````

## 25. Combining Data - `merge()`

`merge()` combines DataFrames using a common key.

### Question 41

Create:

- one DataFrame containing `Student_ID` and `Name`,
- another containing `Student_ID` and `Marks`.

Then merge them using `Student_ID`.


````python
df_1 = pd.DataFrame({
    "Student_ID": [1,2,3,4,5],
    "Name": ["A", "B", "C", "D", "E"],
})

df_2 = pd.DataFrame({
    "Student_ID": [1,2,3,4,5],
    "Marks": [100,80,40,70,80]
})

pd.merge(df_1,df_2)
````

## 26. Combining Data with `concat()`

`concat()` joins DataFrames along an axis.

### Question 42

Concatenate two DataFrames vertically.


````python
df_1 = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Marks": [100,80,40,70,80]
})

df_2 = pd.DataFrame({
    "Name": ["F", "G", "H", "I", "J"],
    "Marks": [100,80,40,70,80],
    "Age": [20,21,19, np.nan, 21]
})

pd.concat([df_1, df_2], ignore_index=True)
````

For vertical concatenation:

```python
pd.concat([df1, df2], ignore_index=True)
```

- **`merge()`** combines tables using a common key, similar to joining database tables.
- **`concat()`** stacks DataFrames vertically or horizontally.

## 27. Machine Learning
