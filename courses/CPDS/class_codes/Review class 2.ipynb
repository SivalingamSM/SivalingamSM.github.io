```python
f = [[0,3,4,5],
           [0,3,4,5],
           [0,3,0,5],
           [1,3,4,5]]
```

```python
plt.imshow(f, cmap = "gray",vmin=0, vmax = 5)
```

```python
g = [[1,1],
    [2,2]]
```

```python
f
```

```python
for i in range(len(f)-1):
  for j in range(len(f[0])-1):
    patch = [f[i][j:j+2],
             f[i+1][j:j+2]]
    print(patch)
```

```python
convlutedimg = []
for i in range(len(f)-1):
  row = []
  for j in range(len(f[0])-1):
    sum = 0
    for ki in range(2):
      for kj in range(2):
        sum = sum + f[i+ki][j+kj]*g[ki][kj]
    row.append(sum/4)
  convlutedimg.append(row)
```

```python
convlutedimg
```

```python
plt.imshow(f,cmap = "grey")
plt.show()
plt.imshow(convlutedimg,cmap = "grey")
plt.show()
```

```python
g = [[-1,-1,-1],
     [-1,-1,-1],
     [-1,-1,-1]]
```

```python
convlutedimg = []
for i in range(399-2):
  row = []
  for j in range(399-2):
    sum = 0
    for ki in range(3):
      for kj in range(3):
        sum = sum + image.getpixel((i+ki,j+kj))*g[ki][kj]
    row.append(sum/9)
  convlutedimg.append(row)
```
