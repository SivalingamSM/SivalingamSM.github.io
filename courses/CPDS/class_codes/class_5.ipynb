```python
N = int(input("Enter N: "))

for i in range(1, N+1):
    print(2*i, end=" ")
print()

for i in range(1, N+1):
    print(2*i-1, end=" ")
print()

for i in range(1, N+1):
    print(i**2, end=" ")
print()

for i in range(1, N+1):
    print(i**3, end=" ")
print()

start = int(input("Range start: "))
end = int(input("Range end: "))
for n in range(start, end+1):
    if n > 1:
        prime = True
        for d in range(2, int(n**0.5)+1):
            if n % d == 0:
                prime = False
                break
        if prime:
            print(n, end=" ")
print()
```
