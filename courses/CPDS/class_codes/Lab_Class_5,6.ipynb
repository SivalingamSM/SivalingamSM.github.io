## 32. A Graph Can Be Correct but Misleading

Consider the marks:

```python
subjects = ["Math", "Physics", "Chemistry", "Python"]
marks = [88, 91, 89, 93]
```


````python
import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "Chemistry", "Python"]
marks = [88, 91, 89, 93]

plt.bar(subjects, marks)

plt.show()

import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "Chemistry", "Python"]
marks = [88, 91, 89, 93]

plt.bar(subjects, marks)
plt.ylim(85, 95)
plt.show()
````
